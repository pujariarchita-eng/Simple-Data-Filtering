# Simple Data Filtering

**Objective:** Create student records and filter students based on marks, age, or department using Python conditional statements.


```python
# Student records represented as a list of dictionaries
students = [
    {'name': 'Aarav', 'marks': 85, 'age': 20, 'department': 'ENTC'},
    {'name': 'Priya', 'marks': 72, 'age': 21, 'department': 'Computer'},
    {'name': 'Rahul', 'marks': 91, 'age': 20, 'department': 'ENTC'},
    {'name': 'Sneha', 'marks': 64, 'age': 22, 'department': 'Mechanical'},
    {'name': 'Aditya', 'marks': 78, 'age': 19, 'department': 'Computer'},
    {'name': 'Neha', 'marks': 88, 'age': 21, 'department': 'ENTC'},
    {'name': 'Rohan', 'marks': 55, 'age': 20, 'department': 'Civil'},
    {'name': 'Kavya', 'marks': 95, 'age': 19, 'department': 'Computer'}
]

print('All Student Records:')
for student in students:
    print(student)
```

```python
# Filter 1: Students having marks greater than or equal to 80
high_scorers = [student for student in students if student['marks'] >= 80]

print('Students with marks >= 80:')
for student in high_scorers:
    print(student)
```

```python
# Filter 2: Students from ENTC department and marks >= 80
entc_high_scorers = [
    student for student in students
    if student['department'] == 'ENTC' and student['marks'] >= 80
]

print('ENTC students with marks >= 80:')
for student in entc_high_scorers:
    print(student)
```

```python
# Filter 3: Students younger than 21 years and marks >= 75
young_good_students = [
    student for student in students
    if student['age'] < 21 and student['marks'] >= 75
]

print('Students younger than 21 with marks >= 75:')
for student in young_good_students:
    print(student)
```

## Optional Pandas DataFrame Filtering

The same records can be converted into a Pandas DataFrame and filtered using conditions.

```python
import pandas as pd

df = pd.DataFrame(students)
print(df)
```

```python
# Pandas: marks >= 80
print(df[df['marks'] >= 80])

# Pandas: ENTC department AND marks >= 80
print(df[(df['department'] == 'ENTC') & (df['marks'] >= 80)])
```

## Interview Questions

**1. What is data filtering?**  
Data filtering is the process of selecting only the records that satisfy a specific condition from a dataset.

**2. How would you filter a Pandas DataFrame?**  
A Pandas DataFrame can be filtered using Boolean conditions. For example: `df[df['marks'] >= 80]`.

**3. What is the difference between filtering and sorting?**  
Filtering selects records that meet a condition, while sorting changes the order of records based on one or more columns.
